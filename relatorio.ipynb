# Análise de Dados: Potencial de Retorno de Personagens em Genshin Impact

## 1. Introdução

O mercado de jogos digitais gratuitos (free-to-play) possui diferentes estratégias de monetização. Entre elas, destaca-se o sistema de gacha, utilizado em jogos como Genshin Impact, desenvolvido pela empresa HoYoverse. Nesse modelo, os jogadores utilizam moedas virtuais para obter personagens e equipamentos por meio de um sistema de sorteio, conhecido como banner.

Uma das principais características desse modelo de negócio é a disponibilização periódica de personagens, incluindo o retorno de personagens anteriormente lançados, em eventos conhecidos pela comunidade como reruns. A escolha desses retornos representa uma decisão estratégica para a empresa, uma vez que diferentes personagens podem apresentar níveis distintos de interesse e demanda por parte dos jogadores.

Diante desse cenário, este trabalho propõe uma análise de dados voltada à identificação de personagens com maior potencial para retornar em banners futuros de Genshin Impact. O estudo utiliza informações históricas de personagens, incluindo características individuais, datas de lançamento, histórico de banners e indicadores de comportamento dos jogadores.

A base de dados utilizada foi desenvolvida por uma comunidade de jogadores e reúne informações de 103 personagens disponíveis até a versão 6.2 do jogo. Embora não seja uma base oficial da HoYoverse, ela apresenta dados reais relacionados aos personagens e registros de participação da comunidade.

A partir dessas informações, pretende-se aplicar técnicas de análise exploratória, tratamento de dados e engenharia de atributos utilizando Python e suas principais bibliotecas para Ciência de Dados. Também serão desenvolvidos indicadores derivados, incluindo estimativas de investimento dos jogadores, respeitando as limitações dos dados disponíveis.

Ao final, será elaborado um ranking de potencial de retorno dos personagens, considerando os indicadores identificados durante a análise. Os resultados serão utilizados para formular uma previsão e posteriormente compará-la com os banners efetivamente disponibilizados pela HoYoverse após o período analisado.

## 2. Objetivos

### 2.1 Objetivo geral

Identificar, por meio da análise de dados históricos e indicadores de comportamento dos jogadores, quais personagens de Genshin Impact apresentam maior potencial para retornar em banners futuros, considerando o cenário existente até a versão 6.2 do jogo.

### 2.2 Objetivos específicos

- Explorar e compreender a estrutura da base de dados utilizada.
- Identificar e tratar possíveis inconsistências, valores ausentes e duplicidades.
- Desenvolver novas variáveis que contribuam para a análise do histórico e da demanda dos personagens.
- Investigar a relação entre características dos personagens e os indicadores de interesse da comunidade.
- Analisar a influência do tempo desde o último banner na possibilidade de retorno.
- Desenvolver indicadores para estimar o investimento monetário associado aos registros de pulls da comunidade.
- Construir um ranking de personagens com maior potencial de retorno.
- Desenvolver uma interface interativa utilizando Streamlit para apresentar os resultados.
- Comparar as previsões obtidas com os banners efetivamente lançados após o período analisado.
