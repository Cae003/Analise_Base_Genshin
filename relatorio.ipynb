# Análise Histórica dos Banners de Genshin Impact

**Disciplina:** Python para Machine Learning e Inteligência Artificial  
**Projeto Final — Análise Exploratória de Dados**

## 1. Introdução ao Problema

Genshin Impact é um jogo que disponibiliza personagens por períodos limitados, conhecidos como *banners*. Após esse período, os personagens podem retornar em atualizações futuras, chamadas de *reruns*, mas não existe um calendário fixo que determine quando isso acontecerá.

O objetivo deste projeto é **analisar o histórico de banners para identificar padrões de ausência e frequência de retorno**, buscando compreender quais personagens apresentam características que os tornam possíveis candidatos a futuras aparições.

Como aplicação prática, a análise utiliza a versão 6.4 (Luna V) como referência.

## 2. Como Executar o Projeto

O projeto foi desenvolvido em Python e utiliza o Streamlit para apresentar os resultados em uma interface interativa.

Para executar a aplicação:

1. Baixe ou clone o repositório do projeto.
2. Instale as bibliotecas necessárias: Streamlit, Pandas, NumPy e Plotly.
3. Certifique-se de que o arquivo `arquivo.csv` está na mesma pasta que o arquivo `main.py`.
4. Abra o terminal na pasta principal do projeto e execute o comando:

   `streamlit run main.py`

5. A aplicação será aberta automaticamente no navegador, normalmente no endereço `http://localhost:8501`.

## 3. Metodologia

### 3.1. Obtenção dos dados

A base de dados utilizada foi obtida na plataforma **Kaggle**, por meio do conjunto de dados *Genshin Impact*, disponibilizado no endereço:

https://www.kaggle.com/datasets/bisheshkhanalcs26/genshin-impact

O conjunto contém informações dos personagens, como raridade, elemento, região, data de lançamento, histórico de banners e indicadores de interesse dos jogadores.

### 3.2. Tratamento dos dados

O tratamento foi realizado em Python, utilizando principalmente as bibliotecas **Pandas e NumPy**.

As principais etapas foram:

- Carregamento e organização dos dados.
- Conversão de datas e valores numéricos.
- Tratamento de registros ausentes ou incompatíveis com determinados cálculos.
- Organização cronológica dos banners.
- Criação de indicadores, como dias desde a última aparição, intervalo médio entre banners e índice de ausência relativa.

### 3.3. Visualização

Para apresentar os resultados, foi desenvolvido um dashboard em **Streamlit**, com gráficos interativos produzidos principalmente com Plotly.

A aplicação foi dividida em três análises: ausência e cronologia, frequência de retorno e interesse dos jogadores.

## 5. Conclusão

A análise demonstrou que os dados históricos permitem identificar padrões de ausência e frequência de retorno dos personagens de Genshin Impact.

O principal resultado foi a criação de indicadores que possibilitam comparar personagens com históricos diferentes, especialmente o **índice de ausência relativa**, que considera o tempo sem aparição em relação ao comportamento anterior de cada personagem.

Com essas informações, é possível elaborar uma lista fundamentada de possíveis candidatos para a versão 6.4, embora não seja possível garantir quais personagens serão escolhidos pela desenvolvedora.

O projeto permitiu aplicar técnicas de preparação, transformação e visualização de dados em Python, além de desenvolver uma interface interativa para facilitar a interpretação dos resultados.
